# Chriesibaum's EMC Calc & Cheat Sheet

```python
import numpy as np
from pprint import pprint

from IPython.display import display, Markdown
import pint
u = pint.UnitRegistry()
u.formatter.default_format = '~P'

def print_md(text):
    display(Markdown(text))
```

```python
f_base = 24 * u.MHz
f_base = 96 * u.MHz

n_harmonics = 10

f = f_base *4 * 2
f = f_base * np.arange(1, n_harmonics + 1)

λ = u.speed_of_light / f

λ_4 = λ / 4

# print a table of harmonic, f, λ and λ/4
table = '| harmonic | f | λ | λ/4 |\n|-|-|-|-|\n'
for i, (f_i, λ_i, λ_4_i) in enumerate(zip(f, λ, λ_4), start=1):
    table += f'| {i} | {f_i:.2f~P} | {λ_i.to("mm"):.2f~P} | {λ_4_i.to("mm"):.2f~P} |\n'
print_md(table)
```
